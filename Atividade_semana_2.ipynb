```python
import matplotlib.pyplot as plt
#definindo a classe livros
class livros():
    def __init__(self, titulo, autor, genero, quantidade):
        self.titulo = titulo
        self.autor = autor
        self.genero = genero
        self.quantidade = quantidade
#criando a lista usando uma lista chamada[livros cadastrados]
livros_cadastrados = []

#comando criado para cadastrar livros
def cadastrar_livro (titulo, autor, genero, quantidade):
  #convertendo a quantidade em numeros inteiros para nao ocorrer erros usando uma str
    livro = livros(titulo, autor, genero, int(quantidade))
    livros_cadastrados.append(livro)
    print(f"livro {titulo} cadastrado com sucesso")

#comando listar livros
def listar_livros():
    for livro in livros_cadastrados:
        print(f"Título: {livro.titulo}, Autor: {livro.autor}, Gênero: {livro.genero},quantidade: {livro.quantidade}")

#comando buscar livros
def buscar_livro(titulo):
    for livro in livros_cadastrados:
        if livro.titulo == titulo:
            print(f"Título: {livro.titulo}, Autor: {livro.autor}, Gênero: {livro.genero},quantidade: {livro.quantidade}")
            return
    print(f"Livro {titulo} não encontrado")

#usando comando input() para poder receber os valores dos livros em seus cadastros
livro1 = cadastrar_livro(input('Digite o título: '),
                        input('Digite o autor: '),
                        input('Digite o gênero: '),
                        input('Digite a quantidade: ') )


#dados titulo, autor, genero, disponivel
genero_lista = ['terror', 'romance', 'ficção', 'aventura', 'drama']

#iniciando a quantidade de generos em 0 na lista
quantidade_generos_por_tipo = {g: 0 for g in genero_lista}

# agregandos os livros cadastrados livros_cadastrados
for livro_obj in livros_cadastrados:
    if livro_obj.genero in quantidade_generos_por_tipo:
        quantidade_generos_por_tipo[livro_obj.genero] += livro_obj.quantidade

#preparando o grafico
generos_para_grafico = list(quantidade_generos_por_tipo.keys())
quantidades_para_grafico = list(quantidade_generos_por_tipo.values())
#criar um grafico de generos de livros cadastrados
plt.bar(generos_para_grafico, quantidades_para_grafico)
#rotulos
plt.xlabel('genero')
plt.ylabel('Quantidade(em unidades)')
#titulo do grafico
plt.title('Quantidade de livros por gênero')
#mostrar grafico
plt.show()
#livros
print('livros cadastrados')
listar_livros()
#buscar livros
print("buscar livros")
buscar_livro(input('Digite o titulo: '))
```
